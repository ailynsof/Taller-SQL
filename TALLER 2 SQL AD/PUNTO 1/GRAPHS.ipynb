```python
# =========================
# 1. Conectar Google Drive
# =========================
from google.colab import drive
drive.mount('/content/drive')

file_path = "/content/drive/MyDrive/academic_data.csv"


# =========================
# 2. Librerías
# =========================
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (12,6)


# =========================
# 3. Cargar datos
# =========================
df = pd.read_csv(file_path)

df = df.dropna(subset=["MARKS"])


# =====================================================
# GRAFICA 1 — Promedio de notas por curso
# =====================================================
avg_marks = df.groupby("COURSE_NAME")["MARKS"].mean().sort_values().reset_index()

plt.figure()
plt.bar(avg_marks["COURSE_NAME"], avg_marks["MARKS"])
plt.xticks(rotation=45, ha="right")
plt.title("Average Marks by Course", fontsize=16)
plt.xlabel("Course", fontsize=12)
plt.ylabel("Average Marks", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# =====================================================
# GRAFICA 2 — Distribución de notas
# =====================================================
plt.figure()
plt.hist(df["MARKS"], bins=10)
plt.title("Distribution of Student Marks", fontsize=16)
plt.xlabel("Marks", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# =====================================================
# GRAFICA 3 — Estudiantes por curso
# =====================================================
students_per_course = df.groupby("COURSE_NAME")["STUDENT_ID"].nunique().sort_values().reset_index()

plt.figure()
plt.bar(students_per_course["COURSE_NAME"], students_per_course["STUDENT_ID"])
plt.xticks(rotation=45, ha="right")
plt.title("Students per Course", fontsize=16)
plt.xlabel("Course", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# =====================================================
# GRAFICA 4 — Asistencia vs notas
# =====================================================
plt.figure()
plt.scatter(df["NO_OF_DAYS_OFF"], df["MARKS"])
plt.title("Attendance vs Marks", fontsize=16)
plt.xlabel("Days Off", fontsize=12)
plt.ylabel("Marks", fontsize=12)
plt.grid(alpha=0.4)
plt.tight_layout()
plt.show()


# =====================================================
# GRAFICA 5 — Cursos con bajo rendimiento
# =====================================================
low_perf = df[df["MARKS"] < 70]

low_perf_count = low_perf.groupby("COURSE_NAME").size().sort_values().reset_index(name="LOW_STUDENTS")

plt.figure()
plt.bar(low_perf_count["COURSE_NAME"], low_perf_count["LOW_STUDENTS"])
plt.xticks(rotation=45, ha="right")
plt.title("Courses with Low Performance Students", fontsize=16)
plt.xlabel("Course", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# =====================================================
# GRAFICA 6 — Promedio por elegibilidad de examen
# =====================================================
avg_by_eligibility = df.groupby("ELIGIBILITY_FOR_EXAMS")["MARKS"].mean().reset_index()

plt.figure()
plt.bar(avg_by_eligibility["ELIGIBILITY_FOR_EXAMS"], avg_by_eligibility["MARKS"])
plt.title("Average Marks by Exam Eligibility", fontsize=16)
plt.xlabel("Eligibility for Exams", fontsize=12)
plt.ylabel("Average Marks", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# =====================================================
# GRAFICA 7 — Promedio por año de registro
# =====================================================
avg_by_year = df.groupby("STUDENT_REG_YEAR")["MARKS"].mean().reset_index()

plt.figure()
plt.plot(avg_by_year["STUDENT_REG_YEAR"], avg_by_year["MARKS"], marker="o")
plt.title("Average Marks by Student Registration Year", fontsize=16)
plt.xlabel("Registration Year", fontsize=12)
plt.ylabel("Average Marks", fontsize=12)
plt.grid(alpha=0.4)
plt.tight_layout()
plt.show()


# =====================================================
# GRAFICA 8 — Número de exámenes por curso
# =====================================================
exams_per_course = df.groupby("COURSE_NAME")["EXAM_NAME"].nunique().sort_values().reset_index()

plt.figure()
plt.bar(exams_per_course["COURSE_NAME"], exams_per_course["EXAM_NAME"])
plt.xticks(rotation=45, ha="right")
plt.title("Number of Exams per Course", fontsize=16)
plt.xlabel("Course", fontsize=12)
plt.ylabel("Number of Exams", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# =========================
# 1. Conectar Google Drive
# =========================
from google.colab import drive
drive.mount('/content/drive')

file_path = "/content/drive/MyDrive/academic_data.csv"


# =========================
# 2. Librerías
# =========================
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
plt.rcParams["figure.figsize"] = (12,6)


# =========================
# 3. Cargar datos
# =========================
df = pd.read_csv(file_path)
df = df.dropna(subset=["MARKS"])


# =====================================================
# 1. Promedio de notas por curso
# =====================================================
avg_marks = df.groupby("COURSE_NAME")["MARKS"].mean().sort_values().reset_index()

plt.figure()
plt.bar(avg_marks["COURSE_NAME"], avg_marks["MARKS"], color="skyblue", edgecolor="black")
plt.xticks(rotation=45, ha="right")
plt.title("Average Marks by Course", fontsize=16, fontweight="bold")
plt.xlabel("Course")
plt.ylabel("Average Marks")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()


# =====================================================
# 2. Distribución de notas
# =====================================================
plt.figure()
plt.hist(df["MARKS"], bins=10, color="lightcoral", edgecolor="black")
plt.title("Distribution of Student Marks", fontsize=16, fontweight="bold")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()


# =====================================================
# 3. Asistencia vs notas
# =====================================================
plt.figure()
plt.scatter(df["NO_OF_DAYS_OFF"], df["MARKS"], color="purple", alpha=0.6)
plt.title("Attendance vs Marks", fontsize=16, fontweight="bold")
plt.xlabel("Days Off")
plt.ylabel("Marks")
plt.grid(alpha=0.4)
plt.tight_layout()
plt.show()


# =====================================================
# 4. Cursos con bajo rendimiento
# =====================================================
low_perf = df[df["MARKS"] < 70]
low_perf_count = low_perf.groupby("COURSE_NAME").size().sort_values().reset_index(name="LOW_STUDENTS")

plt.figure()
plt.bar(low_perf_count["COURSE_NAME"], low_perf_count["LOW_STUDENTS"], color="orange", edgecolor="black")
plt.xticks(rotation=45, ha="right")
plt.title("Courses with Low Performance Students", fontsize=16, fontweight="bold")
plt.xlabel("Course")
plt.ylabel("Number of Students")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()


# =====================================================
# 5. Elegibilidad vs notas
# =====================================================
avg_by_eligibility = df.groupby("ELIGIBILITY_FOR_EXAMS")["MARKS"].mean().reset_index()

plt.figure()
plt.bar(avg_by_eligibility["ELIGIBILITY_FOR_EXAMS"], avg_by_eligibility["MARKS"], color="teal", edgecolor="black")
plt.title("Average Marks by Exam Eligibility", fontsize=16, fontweight="bold")
plt.xlabel("Eligibility for Exams")
plt.ylabel("Average Marks")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
